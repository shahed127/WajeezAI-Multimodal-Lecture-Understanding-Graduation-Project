```python
from google import genai

# The client gets the API key from the environment variable `GEMINI_API_KEY`.
client = genai.Client(api_key=secret_value_0)

tts_postproc_prompt = f"""
You are an expert post-processing editor specialized in Arabic with natural English code-switching.

Your ONLY job is to clean and polish raw ASR transcriptions while staying extremely faithful to the spoken content.

Rules:
- Keep all English words and code-switches exactly as spoken — never translate or remove them.
- Fix obvious ASR errors (wrong words, missing/repeated syllables, bad spelling).
- Add natural punctuation, capitalization, and sentence breaks for readability.
- Improve flow between sentences but NEVER change meaning or add information.
- convert to Fus'ha (Modern Standard Arabic).
- Output ONLY the cleaned transcription. No explanations, no notes, no markdown.

The ASR transcription:
{ASR_transcription}
"""

response = client.models.generate_content(
    model="gemini-3.1-flash-lite-preview", contents=tts_postproc_prompt
)
print(response.text)
```
